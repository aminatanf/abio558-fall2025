```python
import pandas as pd 
import glob
```

```python
folder_path = '/Users/aminatafofana/Downloads/GSE121780_RAW/'
```

```python
file_list = glob.glob(folder_path + "*genes.both.count.txt.gz")
print("Files found:", file_list)
```

```python


dfs = []

for file in file_list:
    sample_name = file.split("/")[-1].split("_")[0]  # GSM ID

    # skip the first row (BAM file names), header=None treats first row as data
    df = pd.read_csv(file, sep="\t", compression="gzip", skiprows=1, header=None)

    # first column = gene, last column = counts
    df = df.iloc[:, [0, -1]]
    df.columns = ["Gene", sample_name]
    df.set_index("Gene", inplace=True)

    dfs.append(df)

# Combine all samples
data = pd.concat(dfs, axis=1)

print(data.shape)
print(data.head())



```
